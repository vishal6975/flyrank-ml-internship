## 3. Leakage audit

*The same hunt from Week 3, on your final feature set.*

I reviewed the final feature set based on the question: Would this information genuinely be available when the prediction is made?

The target future_ctr_declined is created using late-March CTR compared with early-March CTR. The late-March CTR represents future outcome information and is not included as a model feature.

The model uses impressions_early, clicks_early, ctr_early, and avg_position_early. These features are calculated from the early period and occur before the late-period outcome.

The main issue identified in the Week-5 model was evaluation leakage: the model was trained and evaluated on the same observations. The Week-6 grouped-client split addresses this by evaluating the model on clients that were not used during training.

I also checked that clients do not overlap between the training and validation groups.

Failure examples

I inspected false positives and false negatives from the held-out client validation set. These examples show that the model does not perfectly distinguish observations with a later CTR decline from those without one.

The observed errors support using the model as directional decision-support rather than as a guaranteed prediction system.

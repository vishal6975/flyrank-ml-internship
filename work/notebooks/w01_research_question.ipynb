## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

Looking at the starter dataset: 54.2% of pages are currently declining — more than half the inventory. Declining pages (avg 4,919 impressions) actually have slightly higher visibility than growing pages (avg 4,716 impressions), meaning decline isn't simply about low-traffic pages — there's real signal worth investigating.

```python
import pandas as pd
df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

declining_pct = (df["trend_direction"] == "down").mean() * 100
print(f"% of pages currently declining: {declining_pct:.1f}%")

avg_impressions_declining = df[df["trend_direction"]=="down"]["impressions_90d"].mean()
avg_impressions_up = df[df["trend_direction"]=="up"]["impressions_90d"].mean()
print(f"Avg impressions (declining): {avg_impressions_declining:.0f}")
print(f"Avg impressions (growing): {avg_impressions_up:.0f}")
```

## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

My lane is Refresh/Content Opportunity Scoring. This is a classification/scoring
task — I'm ranking pages by likelihood of needing review, using a binary or
probability-based classifier to score each page, then sorting the ranked list.
It's not pure clustering or regression; it's closest to a scoring/ranking
classification problem.

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

Proxy target: is_declining_label = trend_direction == "down" (the starter dataset's
existing proxy). This is a reasonable starting proxy but not ideal — a stronger
target for my capstone would be a future-looking outcome, like "will this page
decline over the next 30 days," to avoid just describing the current state.

## 3. Success metric

*One metric you can defend. What number means 'good'?*

Precision@50 — because a content reviewer can only realistically review a limited
number of pages per week (e.g. top 50). This metric asks: of the top 50 ranked
pages, how many are actually genuinely declining? This matches the real decision
capacity better than overall accuracy.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

The starter notebooks already showed this empirically: a hand-written rule scored
Precision@50 of 0.240, while a random forest model scored 0.740 — roughly 3x better.
A fixed rule can only combine a few thresholds a human guessed at, but a learned
model can find non-obvious combinations across many signals that a human wouldn't
manually encode. With thousands of pages and limited reviewer time, even a modest
accuracy improvement translates to real pages saved from being missed.
